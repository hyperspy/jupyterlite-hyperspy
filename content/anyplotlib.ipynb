## AnyPlotlib

Example using [anyplotlib](https://cssfrancis.github.io/anyplotlib/), an interactive
plotting library built on [anywidget](https://anywidget.dev) with a
matplotlib-like object-oriented API. All rendering happens in the browser, so
pan/zoom/drag stays fast even on large data — without a kernel round-trip per
frame.

`anyplotlib` is pre-installed in this deployment, so it can be imported directly
without installing it first.

```python
import numpy as np
import anyplotlib as apl
```

### Plotting a line

Same shape as `plt.subplots(1, 1)`: create a `Figure`, call methods on `Axes`
(note the data goes first — pass the x-axis values with `axes=[...]`).
Try dragging and zooming — it is fully interactive in the browser.

```python
x = np.linspace(0, 10, 1000)

fig, ax = apl.subplots(1, 1, figsize=(700, 350))
ax.plot(np.sin(x), axes=[x], label="sin")
ax.plot(np.cos(x), axes=[x], label="cos")
fig
```

### Images with a crosshair widget

A synthetic hyperspectral dataset: an image plus a spectrum, linked by a
draggable crosshair. `pointer_move` fires on every drag frame (keep it cheap);
`pointer_settled` fires once when the drag ends.

```python
stack = np.random.random((128, 128, 64))
stack[64:, 64:, :] += np.linspace(0, 1, 64)

fig, (ax_img, ax_spec) = apl.subplots(1, 2, figsize=(900, 400))
img = ax_img.imshow(stack.mean(axis=2), cmap="viridis")
spec = ax_spec.plot(stack[64, 64], units="channel")

cross = img.add_widget("crosshair", cx=64, cy=64)

@cross.add_event_handler("pointer_move")
def update(event):
    spec.set_data(stack[int(cross.cy), int(cross.cx)])

fig
```

Drag the crosshair over the image — the spectrum updates live, and the widget
position is synced back to Python:
