## AnyWidget

Simple example using [anywidget](https://anywidget.dev) to create custom Jupyter widgets from a plain Python class.
The frontend code (HTML/JS) is defined inline in the class attributes.

`anywidget` is pre-installed in this deployment, so it can be imported directly
without installing it first.

```python
import anywidget
import traitlets


class CounterWidget(anywidget.AnyWidget):
    """A simple counter widget."""

    count = traitlets.Int(0).tag(sync=True)
    value = traitlets.Float(0.5).tag(sync=True)

    _esm = """
    function render({ model, el }) {
        let button = document.createElement("button");
        button.style.fontSize = "1.2em";
        button.style.margin = "0.5em";
        button.onclick = () => {
            model.set("count", model.get("count") + 1);
            model.save_changes();
        };
        let slider = document.createElement("input");
        slider.type = "range";
        slider.min = 0;
        slider.max = 1;
        slider.step = 0.01;
        slider.style.verticalAlign = "middle";
        slider.oninput = () => {
            model.set("value", parseFloat(slider.value));
            model.save_changes();
        };
        let output = document.createElement("span");
        output.style.verticalAlign = "middle";
        let update = () => {
            output.textContent = `Count: ${model.get("count")}, Value: ${model.get("value").toFixed(2)}`;
        };
        model.on("change", update);
        slider.value = model.get("value");
        update();
        el.appendChild(button);
        el.appendChild(slider);
        el.appendChild(output);
    }
    export default { render };
    """


counter = CounterWidget()
counter
```

The widget state is synced between the frontend and the Python kernel, so the
traits can be read and set from Python:

```python
counter.count, counter.value
```

```python
counter.value = 0.75
```
